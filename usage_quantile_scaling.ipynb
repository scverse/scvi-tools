# Scaling

### `quantile_minmax` Scaling

Scales each feature independently using its lower ($1-q$) and upper ($q$) quantiles, where \($q=0.999$) by default.

The transformation is defined as:

$$
x' = \frac{x - Q_{1-q}(X)}{Q_q(X) - Q_{1-q}(X)}
$$

- **Lower quantile:** Mapped to 0.
- **Upper quantile:** Mapped to 1.
- **`clip_quantile=True`:** Values outside the quantile boundaries are clipped to [0, 1].
- **`clip_quantile=False` (default):** Values outside the quantile boundaries can be below 0 or above 1.

This approach is a more robust version of the min_max scaling, where $q \%$ of the data will end up in the range $[0, 1]$ (with clip_quantile=False) or $100 \%$ (with clip_quantile=True).

```python
cytovi.scale(adata, method="quantile_minmax", clip_quantile=False, quantile=0.999)
```
